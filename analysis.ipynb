```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the data using the REAL Absolute Path
SUMMARY_PATH = "/mnt/upschrimpf2/scratch/mahdipou/models/Anhedonic-AI/Experiment1/evaluation/scienceQA/results/summary.csv"

if not os.path.exists(SUMMARY_PATH):
    raise FileNotFoundError(f"Error: Could not find {SUMMARY_PATH}")

df = pd.read_csv(SUMMARY_PATH)

tiers = df['tier'].tolist()
x = np.arange(len(tiers))
width = 0.35

# -------------------------------------------------------------------------
# Plot 1: Cognitive Capacity Drop (Accuracy)
# -------------------------------------------------------------------------
fig1, ax1 = plt.subplots(figsize=(8, 6))
acc_values = df['acc_%'].fillna(0).values

colors1 = ['#4C72B0', '#C44E52']
bars1 = ax1.bar(tiers, acc_values, color=colors1, width=0.5, edgecolor='black')

ax1.set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
ax1.set_title('Impact of Ablation on Cognitive Accuracy', fontsize=14, fontweight='bold')
ax1.set_ylim(0, max(acc_values) + 15)

for bar in bars1:
    height = bar.get_height()
    ax1.annotate(f'{height:.1f}%',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# Plot 2: Attempted vs Earned Points (Reward Seeking vs Actual Performance)
# -------------------------------------------------------------------------
fig2, ax2 = plt.subplots(figsize=(10, 6))

att_pts = df['mean_att'].fillna(0).values
ern_pts = df['mean_earn'].fillna(0).values

rects1 = ax2.bar(x - width/2, att_pts, width, label='Attempted Points (Greed)', color='#55A868', edgecolor='black')
rects2 = ax2.bar(x + width/2, ern_pts, width, label='Earned Points (Reality)', color='#DD8452', edgecolor='black')

ax2.set_ylabel('Average Points', fontsize=12, fontweight='bold')
ax2.set_title('Attempted vs. Earned Points per Tier', fontsize=14, fontweight='bold')
ax2.set_xticks(x)
ax2.set_xticklabels(tiers, fontsize=12, fontweight='bold')
ax2.legend(fontsize=11)
ax2.set_ylim(0, max(att_pts) + 15)

for rects in [rects1, rects2]:
    for rect in rects:
        height = rect.get_height()
        ax2.annotate(f'{height:.1f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# Plot 3: Distribution of Question Choices
# -------------------------------------------------------------------------
fig3, ax3 = plt.subplots(figsize=(10, 6))

p1   = df['1pt_att_%'].fillna(0).values
p10  = df['10pt_att_%'].fillna(0).values
p50  = df['50pt_att_%'].fillna(0).values
p100 = df['100pt_att_%'].fillna(0).values

ax3.bar(tiers, p1,   label='1 pt',   color='#EAEAF2', edgecolor='black', width=0.5)
ax3.bar(tiers, p10,  label='10 pts', color='#B3C3DF', edgecolor='black', width=0.5, bottom=p1)
ax3.bar(tiers, p50,  label='50 pts', color='#718BB7', edgecolor='black', width=0.5, bottom=p1+p10)
ax3.bar(tiers, p100, label='100 pts',color='#314D79', edgecolor='black', width=0.5, bottom=p1+p10+p50)

ax3.set_ylabel('Percentage of Choices (%)', fontsize=12, fontweight='bold')
ax3.set_title('Distribution of Selected Question Difficulties', fontsize
```
